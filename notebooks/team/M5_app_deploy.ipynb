## เชื่อม Drive และเตรียมที่เก็บงาน

สร้างโฟลเดอร์เมื่อยังไม่มี และเตรียมฟังก์ชันคัดลอกผลจาก runtime ไป Drive โดยไม่เขียนทับไฟล์ชื่อเดิม.

```python
from pathlib import Path
from shutil import copy2
from google.colab import drive

drive.mount('/content/drive')
PROJECT_ROOT = Path('/content/drive/MyDrive/Bicycle_Rental_Project_Team')
if not PROJECT_ROOT.is_dir():
    raise FileNotFoundError('Add the shared Bicycle_Rental_Project_Team folder as a shortcut in MyDrive first: https://drive.google.com/drive/folders/1dvHp15dXx96rzKizH9NF2bMUg_MBQvv1')
SHARED_DIR = PROJECT_ROOT / '00_SHARED'
DATA_DIR = SHARED_DIR / 'dataset_from_lecturer'
MEMBER_DIR = PROJECT_ROOT / 'M5_APP_DEPLOY'
NOTEBOOKS_DIR = MEMBER_DIR / 'notebooks'
OUTPUTS_DIR = MEMBER_DIR / 'outputs'
HANDOFF_DIR = MEMBER_DIR / 'handoff'
for folder in (DATA_DIR, SHARED_DIR/'shared_notes', SHARED_DIR/'final_artifacts', NOTEBOOKS_DIR, OUTPUTS_DIR, HANDOFF_DIR):
    folder.mkdir(parents=True, exist_ok=True)
print('โฟลเดอร์งานพร้อมแล้ว:', MEMBER_DIR)

def copy_to_drive(source_path, destination_folder=OUTPUTS_DIR, file_name=None):
    source_path = Path(source_path)
    if not source_path.is_file():
        raise FileNotFoundError(f'ไม่พบไฟล์ต้นทาง: {source_path}')
    destination_folder = Path(destination_folder)
    destination_folder.mkdir(parents=True, exist_ok=True)
    destination = destination_folder / (file_name or source_path.name)
    if destination.exists():
        raise FileExistsError(f'มีไฟล์ชื่อนี้แล้ว กรุณาตั้งชื่อเวอร์ชันใหม่: {destination}')
    copy2(source_path, destination)
    print('บันทึกแล้ว:', destination)
    return destination
```

```python
from pathlib import Path
from google.colab import drive
import pandas as pd

drive.mount('/content/drive')
PROJECT_ROOT = Path('/content/drive/MyDrive/Bicycle_Rental_Project_Team')
if not PROJECT_ROOT.is_dir():
    raise FileNotFoundError('Add the shared team folder as a shortcut in MyDrive: https://drive.google.com/drive/folders/1dvHp15dXx96rzKizH9NF2bMUg_MBQvv1')
DATA_PATH = PROJECT_ROOT / '00_SHARED/dataset_from_lecturer/03_Bicycle_Rental_Demand.csv'
if not DATA_PATH.is_file():
    raise FileNotFoundError(f'Missing lecturer CSV: {DATA_PATH}')
df = pd.read_csv(DATA_PATH, encoding='utf-8-sig', parse_dates=['timestamp'])
"""Shared Drive paths and report evidence helpers; no ML exercises are implemented."""

import base64
import hashlib
import importlib.metadata
import json
import platform
import shutil
import sys
import time
import uuid
from datetime import datetime, timezone
from pathlib import Path
WORKFLOW_STAGES = {'S01': {'id': 'S01', 'role': 'M1', 'title': 'ตรวจข้อมูลเบื้องต้นและส่งคำอธิบาย', 'requires': [], 'outputs': ['data_dictionary', 'quality_report'], 'steps': ['รันช่องเชื่อม Drive อ่าน CSV และตั้งค่าหลักฐานรายงาน', 'อ่าน Guide ยืนยัน 1 แถว = 1 ชั่วโมง target rental_count หน่วย rentals/hour', 'ทำหัวข้อ ยืนยันไฟล์และความหมายคอลัมน์ และ ตรวจข้อมูลเบื้องต้น ตรวจ schema/ค่าว่าง/ซ้ำ/ชนิดข้อมูล', 'ใส่ data_dictionary และ quality_report ที่ทำจริงใน OUTPUTS แล้วรัน ส่งต่อ S01'], 'stop': 'หยุดก่อนหัวข้อ สำรวจรูปแบบการเช่า ยังไม่ใช้ทั้งไฟล์ทำ EDA เพื่อเลือก feature/โมเดล', 'resume': 'ให้ M2 ทำ S02 ส่ง train และ split_manifest แล้ว M1 กลับมาทำ S03', 'parallel': ''}, 'S02': {'id': 'S02', 'role': 'M2', 'title': 'แบ่งชุดข้อมูลตามเวลา', 'requires': ['data_dictionary', 'quality_report'], 'outputs': ['train', 'validation', 'test', 'split_manifest'], 'steps': ['รอ S01 และอ่านคำอธิบายที่ M1 ส่ง', 'ทำหัวข้อ ตกลงข้อมูลที่จะใช้ และ เลือกวิธีแบ่งชุด เรียง timestamp แล้วแบ่ง 70%/15%/15% ตาม Guide', 'ตรวจชุดไม่ซ้อนและ train มาก่อน validation มาก่อน test บันทึกจำนวนแถว/ช่วงเวลา/target/features ใน split_manifest', 'ใส่ train validation test split_manifest ใน OUTPUTS แล้วรัน ส่งต่อ S02'], 'stop': 'หยุดก่อนหัวข้อ ออกแบบ preprocessing เพื่อให้ M1 ทำ EDA บน train ก่อน', 'resume': 'รอ eda_observations จาก M1 S03 แล้วกลับมาทำ S04', 'parallel': 'M5 เริ่ม P05 เตรียม UI ได้หลัง S01'}, 'P05': {'id': 'P05', 'role': 'M5', 'title': 'เตรียมหน้ากรอกระหว่างรอโมเดล', 'requires': ['data_dictionary'], 'outputs': [], 'steps': ['อ่าน Guide และ data_dictionary เพื่อกำหนดช่องกรอก หน่วย และ validation', 'ทำหัวข้อ รับไฟล์จากทีม เฉพาะตรวจข้อตกลง และ ออกแบบหน้ากรอกข้อมูล', 'เตรียมโครงแอป/ข้อความผลลัพธ์และจดหลักฐานตามจริง'], 'stop': 'หยุดก่อนหัวข้อ เชื่อมและทดสอบ ยังไม่โหลดโมเดลหรือแสดง prediction สมมติ', 'resume': 'รอ final_pipeline input_schema model_metadata และ final_test_metrics เพื่อเข้า S09', 'parallel': 'ทำขนานกับ S02 ถึง S08 ได้'}, 'S03': {'id': 'S03', 'role': 'M1', 'title': 'กลับมาทำ EDA บน train', 'requires': ['train', 'split_manifest'], 'outputs': ['eda_observations'], 'steps': ['กลับ notebook M1 รันช่อง ตรวจเริ่ม S03 ให้โหลด train เป็น df_eda', 'ทำหัวข้อ สำรวจรูปแบบการเช่า โดยใช้ df_eda ไม่ใช้ raw df ทั้งไฟล์เลือก feature', 'แสดงกราฟพร้อมคำอธิบาย เก็บรูปและสรุปปัญหา/เหตุผลเสนอ preprocessing', 'ใส่ eda_observations เป็นข้อความสรุปจริงแล้วรัน ส่งต่อ S03'], 'stop': 'หยุดหลังส่ง eda_observations ให้ M2 ไม่ปรับแต่ง split/preprocessing แทนขั้นถัดไป', 'resume': 'M2 ทำ S04; M1 เขียนรายงานด้านข้อมูลได้จากหลักฐานรอบนี้', 'parallel': ''}, 'S04': {'id': 'S04', 'role': 'M2', 'title': 'สร้าง preprocessing ร่วม', 'requires': ['train', 'validation', 'split_manifest', 'eda_observations'], 'outputs': ['preprocessing_recipe', 'preprocessor'], 'steps': ['กลับ notebook M2 หลัง ส่งต่อ S02 แล้วรัน ตรวจเริ่ม S04', 'อ่าน EDA จาก M1 ทำหัวข้อ ออกแบบ preprocessing และ ทำ recipe ใช้ร่วมกัน', 'เขียน preprocessor และ fit เฉพาะ train กำหนด feature/order/หน่วยให้ตรงกัน ไม่ fit จาก validation/test', 'ใส่ preprocessing_recipe กับ preprocessor ใน OUTPUTS แล้วรัน ส่งต่อ S04'], 'stop': 'หยุดก่อนขั้น สร้าง final pipeline ยังไม่เลือกโมเดลหรือเปิด test', 'resume': 'รอ M3/M4 ทดลองและส่ง model_selection กับ model_comparison จาก S06 แล้วกลับมาทำ S07', 'parallel': ''}, 'S05A': {'id': 'S05A', 'role': 'M3', 'title': 'baseline และโมเดล A', 'requires': ['train', 'validation', 'split_manifest', 'preprocessing_recipe', 'preprocessor'], 'outputs': ['baseline_pipeline', 'baseline_metrics', 'model_a_pipeline', 'model_a_metrics'], 'steps': ['เริ่มเมื่อ S04 ส่งไฟล์ครบ ตรวจ fingerprint/feature/order ให้ตรงกับข้อตกลง', 'ทำ baseline แล้วทดลองอัลกอริทึม A ด้วย train และ preprocessing ร่วม', 'วัด MAE RMSE R² บน validation จด settings/seed/เหตุผลและข้อจำกัดจากผลจริง', 'ส่ง baseline/A pipeline และ metrics ผ่าน ส่งต่อ S05A ระบุ dataset_split เป็น validation ใน metrics'], 'stop': 'หยุดหลังส่งผล A ไม่เปิด test และไม่ประกาศโมเดลสุดท้ายล่วงหน้า', 'resume': 'ให้ M4 รวมผลใน S06 เมื่อผล B พร้อม', 'parallel': 'ทำขนานกับ S05B'}, 'S05B': {'id': 'S05B', 'role': 'M4', 'title': 'ทดลองโมเดล B', 'requires': ['train', 'validation', 'split_manifest', 'preprocessing_recipe', 'preprocessor'], 'outputs': ['model_b_pipeline', 'model_b_metrics'], 'steps': ['เริ่มเมื่อ S04 ส่งไฟล์ครบ ยังไม่ต้องรอผล A เพื่อฝึก B', 'ทำหัวข้อ ใช้ข้อตกลงชุดเดียวกัน และ ทดลองอัลกอริทึม B', 'ใช้ train/validation/preprocessing เดียวกับ M3 วัด MAE RMSE R² และจด settings/seed', 'ส่ง model_b_pipeline และ model_b_metrics ผ่าน ส่งต่อ S05B'], 'stop': 'หยุดก่อนหัวข้อ เทียบบน validation ถ้ายังไม่มี baseline/A จาก M3', 'resume': 'เมื่อ S05A กับ S05B ส่งครบ ให้ M4 ทำ S06', 'parallel': 'ทำขนานกับ S05A'}, 'S06': {'id': 'S06', 'role': 'M4', 'title': 'รวม validation และล็อกวิธีร่วมกับทีม', 'requires': ['baseline_pipeline', 'baseline_metrics', 'model_a_pipeline', 'model_a_metrics', 'model_b_pipeline', 'model_b_metrics'], 'outputs': ['model_comparison', 'model_selection'], 'steps': ['รัน ตรวจเริ่ม S06 แล้วโหลด metrics ของ baseline/A/B', 'ทำหัวข้อ เทียบบน validation และ อ่าน error บนข้อมูลที่เปิดใช้แล้ว', 'รวมตารางผลจริง ใช้ validation กับเหตุผลด้านความเร็ว/ขนาด/การอธิบายเลือกวิธีร่วมกัน', 'model_selection ต้องมี pipeline_key ระบุ baseline_pipeline หรือ model_a_pipeline หรือ model_b_pipeline พร้อมเหตุผลและ settings ที่ล็อก แล้วรัน ส่งต่อ S06'], 'stop': 'หยุดก่อนประเมิน final test ไม่ใช้ผล test มาตัดสินใจเลือกหรือจูน', 'resume': 'M2 รับ model_selection ไปทำ S07 แล้ว M4 กลับมาทำ S08 เมื่อ final artifact พร้อม', 'parallel': ''}, 'S07': {'id': 'S07', 'role': 'M2', 'title': 'กลับมาสร้าง final pipeline', 'requires': ['model_selection', 'model_comparison', 'train', 'validation', 'preprocessing_recipe', 'preprocessor'], 'outputs': ['final_pipeline', 'input_schema', 'model_metadata'], 'steps': ['กลับ notebook M2 รัน ตรวจเริ่ม S07 หลัง S06', 'อ่าน pipeline_key จาก model_selection และโหลด candidate ที่ทีมเลือกจริง', 'เขียนขั้นฝึก final pipeline ตามวิธีที่ล็อก หากใช้ train+validation ให้ refit preprocessing ภายใน pipeline จากชุดฝึกสุดท้ายเท่านั้น โดยคง test แยกไว้', 'ส่ง final_pipeline รวม preprocessing พร้อม input_schema และ model_metadata ที่ระบุ input/order/หน่วย/รุ่นไลบรารี แล้วรัน ส่งต่อ S07'], 'stop': 'หยุดหลังส่ง final artifact ไม่จูนวิธีด้วย test', 'resume': 'M4 รับไฟล์ตัวเดียวกันไปประเมิน S08 จากนั้น M5 จึงเชื่อมแอป', 'parallel': ''}, 'S08': {'id': 'S08', 'role': 'M4', 'title': 'กลับมาประเมิน final test', 'requires': ['model_selection', 'final_pipeline', 'input_schema', 'model_metadata', 'test', 'split_manifest'], 'outputs': ['final_test_metrics'], 'steps': ['กลับ notebook M4 รัน ตรวจเริ่ม S08 ซึ่งจะโหลด test ได้ในขั้นนี้', 'โหลด final_pipeline ตัวที่จะใช้จริงและประเมินกับ test ตาม split_manifest', 'บันทึก MAE RMSE R² จำนวนตัวอย่าง hash ของ pipeline และข้อจำกัด ระบุ dataset_split เป็น test', 'ใส่ final_test_metrics แล้วรัน ส่งต่อ S08'], 'stop': 'หยุดหลังประเมิน ห้ามย้อนกลับไปจูนจาก test หากผลไม่น่าพอใจให้รายงานข้อจำกัดและคุยกติกาการประเมินใหม่กับอาจารย์', 'resume': 'M5 ทำ S09 เมื่อไฟล์ final และผล test ครบ', 'parallel': ''}, 'S09': {'id': 'S09', 'role': 'M5', 'title': 'เชื่อมและตรวจแอป', 'requires': ['final_pipeline', 'input_schema', 'model_metadata', 'final_test_metrics'], 'outputs': ['app_source', 'app_test_log'], 'steps': ['กลับ notebook M5 รัน ตรวจเริ่ม S09 ก่อนหัวข้อ เชื่อมและทดสอบ', 'โหลด final_pipeline/schema รุ่นเดียวกัน เขียน input validation และ prediction flow', 'ทำหัวข้อ เทียบ Colab กับแอป ร่วมกับ M4 ทดสอบ input ปกติ ขอบเขต และผิดรูปแบบ พร้อม AppTest ใน Colab', 'ใส่ app_source เป็นข้อความ app.py และ app_test_log เป็น DataFrame ผลจริง แล้วรัน ส่งต่อ S09 โดย app_test_log ต้องมีคอลัมน์ passed เป็น True/False สำหรับแต่ละกรณี'], 'stop': 'หยุดก่อน deploy ถ้า parity หรือกรณีสำคัญยังไม่ผ่าน', 'resume': 'เมื่อบันทึกผลทดสอบครบและแก้ข้อผิดพลาดสำคัญแล้ว ทำ S10', 'parallel': ''}, 'S10': {'id': 'S10', 'role': 'M5', 'title': 'เผยแพร่และเก็บผลใช้งานจริง', 'requires': ['app_source', 'app_test_log', 'final_pipeline', 'input_schema', 'model_metadata'], 'outputs': ['deployment_info'], 'steps': ['ตรวจ app_test_log ว่าผ่านจริงก่อนเริ่มเผยแพร่', 'เตรียม code/config/artifact ตามสิทธิ์แล้ว deploy บน Streamlit Community Cloud', 'เปิด URL จริงตรวจโหลดโมเดลและ prediction flow เก็บ URL รุ่น เวลา ภาพ และข้อจำกัด', 'บันทึก deployment_info ตามสิ่งที่ตรวจจริง และส่งต่อผ่าน ส่งต่อ S10'], 'stop': 'หยุดการอ้างว่า deploy สำเร็จจนกว่า URL จริงจะเปิดและตรวจได้', 'resume': 'M1/M5 เริ่ม monitoring และทั้งทีมรวบรวมชุดรายงาน/สาธิต', 'parallel': 'ทุกบทบาทเขียนรายงานส่วนที่มีหลักฐานได้ตลอด ไม่ต้องรอขั้นนี้เพื่อเริ่มเขียน'}}



class WorkflowPause(RuntimeError):
    """An intentional stop at a documented handoff boundary."""

ROLE_FOLDERS = {'M1': 'M1_DATA_EDA', 'M2': 'M2_SPLIT_PREPROCESS',
                'M3': 'M3_MODEL_A', 'M4': 'M4_MODEL_B_EVALUATION', 'M5': 'M5_APP_DEPLOY'}
ARTIFACT_SPECS = {
    'data_dictionary': ('M1', 'data_dictionary.csv'),
    'quality_report': ('M1', 'quality_report.json'),
    'eda_observations': ('M1', 'eda_observations.md'),
    'train': ('M2', 'train.csv'), 'validation': ('M2', 'validation.csv'),
    'test': ('M2', 'test.csv'), 'split_manifest': ('M2', 'split_manifest.json'),
    'preprocessing_recipe': ('M2', 'preprocessing_recipe.json'),
    'preprocessor': ('M2', 'preprocessor.joblib'),
    'baseline_pipeline': ('M3', 'baseline_pipeline.joblib'),
    'baseline_metrics': ('M3', 'baseline_metrics.json'),
    'model_a_pipeline': ('M3', 'model_a_pipeline.joblib'),
    'model_a_metrics': ('M3', 'model_a_metrics.json'),
    'model_b_pipeline': ('M4', 'model_b_pipeline.joblib'),
    'model_b_metrics': ('M4', 'model_b_metrics.json'),
    'model_comparison': ('M4', 'model_comparison.csv'),
    'model_selection': ('M4', 'model_selection.json'),
    'final_test_metrics': ('M4', 'final_test_metrics.json'),
    'final_pipeline': ('SHARED', 'final_pipeline.joblib'),
    'input_schema': ('SHARED', 'input_schema.json'),
    'model_metadata': ('SHARED', 'model_metadata.json'),
    'app_source': ('M5', 'app.py'), 'app_test_log': ('M5', 'app_test_log.csv'),
    'monitoring_log': ('M5', 'monitoring_log.csv'), 'deployment_info': ('M5', 'deployment_info.json'),
}
UPSTREAM_KEYS = {
    'M1': [],
    'M2': ['data_dictionary', 'quality_report'],
    'M3': ['train', 'validation', 'split_manifest', 'preprocessing_recipe', 'preprocessor'],
    'M4': ['train', 'validation', 'split_manifest', 'preprocessing_recipe', 'preprocessor'],
    'M5': ['data_dictionary'],
}


def _sha(path):
    digest = hashlib.sha256()
    with Path(path).open('rb') as handle:
        for chunk in iter(lambda: handle.read(1024 * 1024), b''):
            digest.update(chunk)
    return digest.hexdigest()


def _json(path, value):
    Path(path).parent.mkdir(parents=True, exist_ok=True)
    Path(path).write_text(json.dumps(value, ensure_ascii=False, indent=2, default=str), encoding='utf-8')


class ReportSession:
    """Capture actual executed cells and outputs in a distinct run folder."""

    def __init__(self, root, role, dataset_path, frame):
        self.root, self.role = Path(root), role
        run_id = datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%SZ') + '_' + uuid.uuid4().hex[:8]
        self.directory = self.root / ROLE_FOLDERS[role] / 'outputs/report_evidence' / run_id
        self.directory.mkdir(parents=True, exist_ok=False)
        self.paths = {}
        for key, (owner, filename) in ARTIFACT_SPECS.items():
            folder = self.root / '00_SHARED/final_artifacts/v1' if owner == 'SHARED' else self.root / ROLE_FOLDERS[owner] / 'handoff/v1'
            self.paths[key] = folder / filename
        self.dataset_sha = _sha(dataset_path)
        versions = {}
        for package in ('pandas', 'numpy', 'scikit-learn', 'joblib', 'matplotlib', 'streamlit'):
            try:
                versions[package] = importlib.metadata.version(package)
            except importlib.metadata.PackageNotFoundError:
                versions[package] = None
        _json(self.directory / 'run_metadata.json', {
            'run_id': run_id, 'role': role, 'started_utc': datetime.now(timezone.utc).isoformat(),
            'python': platform.python_version(), 'packages': versions,
            'dataset_path': str(dataset_path), 'dataset_sha256': self.dataset_sha,
            'note': 'Execution evidence only; does not certify ML correctness or team contributions.'})
        _json(self.directory / 'environment_packages.json', {
            distribution.metadata['Name']: distribution.version
            for distribution in importlib.metadata.distributions()
            if distribution.metadata['Name']})
        _json(self.directory / 'dataset_profile.json', {
            'rows': len(frame), 'columns': list(frame.columns),
            'dtypes': {k: str(v) for k, v in frame.dtypes.items()},
            'missing': {k: int(v) for k, v in frame.isna().sum().items()},
            'duplicate_rows': int(frame.duplicated().sum())})
        _json(self.directory / 'artifact_paths.json', {k: str(v) for k, v in self.paths.items()})
        self.shell = None
        self.cell_number = 0
        self.original_publish = None

    def event(self, kind, **details):
        with (self.directory / 'events.jsonl').open('a', encoding='utf-8') as handle:
            handle.write(json.dumps({'utc': datetime.now(timezone.utc).isoformat(), 'kind': kind,
                                     **details}, ensure_ascii=False, default=str) + '\n')

    def upstream(self):
        result = {}
        for key in UPSTREAM_KEYS[self.role]:
            path = self.paths[key]
            result[key] = {'path': str(path), 'exists': path.is_file()}
            print(('พร้อม' if path.is_file() else 'รอไฟล์จากขั้นก่อนหน้า'), key, ':', path)
        self.event('upstream_check', files=result)
        return result

    def require_stage(self, stage_id):
        stage = WORKFLOW_STAGES[stage_id]
        if stage['role'] != self.role:
            raise ValueError(f'{stage_id} is assigned to {stage["role"]}')
        self._require_files(stage['requires'], stage_id)
        print(f'เริ่ม {stage_id}: {stage["title"]}')
        print('ทำถึงจุดนี้แล้วหยุด:', stage['stop'])
        print('กลับมาทำต่อเมื่อ:', stage['resume'])
        self.event('workflow_stage_started', stage=stage_id)
        loaded = {key: self.load(key) for key in stage['requires']}
        if stage_id == 'S10':
            log = loaded['app_test_log']
            if log.empty or 'passed' not in log or not log['passed'].astype(str).str.strip().str.lower().isin(['true', '1']).all():
                raise WorkflowPause('หยุด S10: app_test_log ต้องมีแถวทดสอบจริงและคอลัมน์ passed เป็น True ทุกแถว ก่อน deploy')
        return loaded

    def _require_files(self, keys, stage_id, own_outputs=False):
        missing = [key for key in keys if not self.paths[key].is_file()]
        if missing:
            details = '\n'.join(f'- {key}: {self.paths[key]}' for key in missing)
            self.event('workflow_stage_blocked', stage=stage_id, missing=missing)
            reason = 'ยังส่งออกผลของช่วงนี้ไม่ครบ' if own_outputs else 'รอไฟล์จากขั้นก่อนหน้า'
            action = ('สร้างผลจริง ใส่ใน OUTPUTS แล้วรันช่องส่งต่อช่วงนี้ใหม่' if own_outputs else
                      'เป็นจุดรอที่ตั้งใจไว้ เมื่อผู้ส่งสร้างไฟล์แล้วให้กลับมารันจุดตรวจนี้ใหม่')
            raise WorkflowPause(f'หยุดที่จุดตรวจ {stage_id}: {reason}\n{details}\n{action}')
        for key in keys:
            path = self.paths[key]
            receipt = path.with_name(path.name + '.provenance.json')
            if receipt.is_file():
                info = json.loads(receipt.read_text(encoding='utf-8'))
                if info.get('dataset_sha256') != self.dataset_sha or info.get('sha256') != _sha(path):
                    raise WorkflowPause(f'หยุด {stage_id}: {key} มีรุ่นข้อมูลหรือเนื้อหาไม่ตรง provenance ให้ผู้ส่งตรวจใหม่')

    def finish_stage(self, stage_id, outputs):
        stage = WORKFLOW_STAGES[stage_id]
        if stage['role'] != self.role:
            raise ValueError(f'{stage_id} is assigned to {stage["role"]}')
        self._require_files(stage['requires'], stage_id)
        self.save_available(outputs, keys=stage['outputs'])
        self._require_files(stage['outputs'], stage_id, own_outputs=True)
        self.event('workflow_stage_handoff', stage=stage_id, outputs=stage['outputs'])
        print(f'ไฟล์ส่งต่อ {stage_id} พร้อมแล้ว ต้องตรวจเนื้อหาตาม checklist ก่อนแจ้งเพื่อน')
        print('หยุด:', stage['stop'])
        print('ขั้นต่อไป:', stage['resume'])

    def load(self, key):
        path = self.paths[key]
        if not path.is_file():
            print('ยังไม่มีไฟล์:', path)
            self.event('input_missing', artifact=key, path=str(path))
            return None
        provenance = path.with_name(path.name + '.provenance.json')
        if provenance.is_file():
            source = json.loads(provenance.read_text(encoding='utf-8'))
            if source.get('dataset_sha256') != self.dataset_sha:
                raise ValueError(f'{key}: dataset fingerprint differs; regenerate upstream artifacts')
        self.event('input_loaded', artifact=key, path=str(path), sha256=_sha(path))
        if path.suffix == '.csv':
            import pandas as pd
            frame = pd.read_csv(path, encoding='utf-8-sig')
            if 'timestamp' in frame:
                frame['timestamp'] = pd.to_datetime(frame['timestamp'])
            return frame
        if path.suffix == '.json':
            return json.loads(path.read_text(encoding='utf-8'))
        if path.suffix == '.joblib':
            import joblib
            return joblib.load(path)  # Only load artifacts created by the trusted team.
        return path.read_text(encoding='utf-8')

    def save(self, key, value):
        if value is None:
            return None
        destination = self.paths[key]
        owner = ARTIFACT_SPECS[key][0]
        if owner != self.role and not (owner == 'SHARED' and self.role == 'M2'):
            raise ValueError(f'{key} belongs to {owner}; current role is {self.role}')
        destination.parent.mkdir(parents=True, exist_ok=True)
        if destination.is_file():
            history = self.directory / 'previous_versions' / (_sha(destination)[:12] + '_' + destination.name)
            history.parent.mkdir(parents=True, exist_ok=True)
            shutil.copy2(destination, history)
        snapshot = self.directory / 'artifacts' / uuid.uuid4().hex[:8] / destination.name
        snapshot.parent.mkdir(parents=True, exist_ok=True)
        if destination.suffix == '.csv':
            value.to_csv(snapshot, index=False, encoding='utf-8-sig')
        elif destination.suffix == '.json':
            _json(snapshot, value)
        elif destination.suffix == '.joblib':
            import joblib
            joblib.dump(value, snapshot)
            if hasattr(value, 'get_params'):
                _json(snapshot.with_name(snapshot.name + '.parameters.json'), {
                    'class': type(value).__name__, 'parameters': value.get_params(deep=True)})
        else:
            snapshot.write_text(str(value), encoding='utf-8')
        shutil.copy2(snapshot, destination)
        receipt = {'artifact': key, 'path': str(destination), 'sha256': _sha(destination),
                   'dataset_sha256': self.dataset_sha, 'role': self.role,
                   'run_directory': str(self.directory), 'saved_utc': datetime.now(timezone.utc).isoformat()}
        _json(destination.with_name(destination.name + '.provenance.json'), receipt)
        self.event('artifact_saved', **receipt)
        print('ส่งต่อแล้ว:', destination)
        return destination

    def save_available(self, namespace, keys=None):
        """Export only objects actually produced under the agreed variable names."""
        allowed = [k for k, (owner, _) in ARTIFACT_SPECS.items()
                   if owner == self.role or (owner == 'SHARED' and self.role == 'M2')]
        if keys is not None:
            allowed = [key for key in allowed if key in keys]
        for key in allowed:
            if key in namespace and namespace[key] is not None:
                self.save(key, namespace[key])
            else:
                print('ยังไม่มีผลลัพธ์ จึงยังไม่สร้างไฟล์:', key)

    def save_figure(self, figure, name):
        safe_name = Path(name).name
        destination = self.directory / 'figures' / (uuid.uuid4().hex[:8] + '_' + safe_name)
        destination.parent.mkdir(parents=True, exist_ok=True)
        figure.savefig(destination, dpi=160, bbox_inches='tight')
        self.event('figure_saved', path=str(destination), sha256=_sha(destination))
        return destination

    def _before_cell(self, info):
        self.cell_number += 1
        self.started = time.perf_counter()
        self.cell_dir = self.directory / 'cells' / f'{self.cell_number:04d}'
        self.cell_dir.mkdir(parents=True, exist_ok=True)
        (self.cell_dir / 'source.py').write_text(info.raw_cell, encoding='utf-8')
        self.previous_stdout, self.previous_stderr = sys.stdout, sys.stderr
        class Tee:
            def __init__(self, original, path):
                self.original, self.path = original, path
            def write(self, text):
                with self.path.open('a', encoding='utf-8') as handle:
                    handle.write(text)
                return self.original.write(text)
            def flush(self):
                self.original.flush()
            def __getattr__(self, key):
                return getattr(self.original, key)
        sys.stdout = Tee(sys.stdout, self.cell_dir / 'stdout.txt')
        sys.stderr = Tee(sys.stderr, self.cell_dir / 'stderr.txt')
        self.event('cell_started', cell=self.cell_number)

    def _after_cell(self, result):
        if not hasattr(self, 'started'):
            return  # The installation cell did not run this session's pre-cell hook.
        sys.stdout, sys.stderr = self.previous_stdout, self.previous_stderr
        if result.result is not None:
            (self.cell_dir / 'result.txt').write_text(repr(result.result), encoding='utf-8')
            if hasattr(result.result, 'to_csv'):
                result.result.to_csv(self.cell_dir / 'result_table.csv', encoding='utf-8-sig')
            if hasattr(result.result, 'savefig'):
                result.result.savefig(self.cell_dir / 'result_figure.png', dpi=160, bbox_inches='tight')
        error = result.error_before_exec or result.error_in_exec
        self.event('cell_finished', cell=self.cell_number, execution_count=result.execution_count,
                   success=error is None, error=None if error is None else str(error),
                   seconds=round(time.perf_counter() - self.started, 3))

    def install_capture(self, shell):
        """Preserve visible output and save rich display evidence without rerunning cells."""
        self.shell = shell
        shell.events.register('pre_run_cell', self._before_cell)
        shell.events.register('post_run_cell', self._after_cell)
        self.original_publish = shell.display_pub.publish
        def publish(data, metadata=None, **kwargs):
            directory = getattr(self, 'cell_dir', self.directory / 'setup_display')
            directory.mkdir(parents=True, exist_ok=True)
            token = uuid.uuid4().hex[:8]
            for mime, extension in [('image/png', 'png'), ('image/jpeg', 'jpg'),
                                    ('image/svg+xml', 'svg'), ('text/html', 'html'), ('text/plain', 'txt')]:
                if mime in data:
                    value = data[mime]
                    path = directory / f'display_{token}.{extension}'
                    if mime in ('image/png', 'image/jpeg'):
                        path.write_bytes(base64.b64decode(value))
                    else:
                        path.write_text(str(value), encoding='utf-8')
            return self.original_publish(data=data, metadata=metadata, **kwargs)
        shell.display_pub.publish = publish
        print('Report evidence:', self.directory)

    def close(self):
        if self.shell:
            self.shell.events.unregister('pre_run_cell', self._before_cell)
            self.shell.events.unregister('post_run_cell', self._after_cell)
            self.shell.display_pub.publish = self.original_publish
            if hasattr(self, 'previous_stdout'):
                sys.stdout, sys.stderr = self.previous_stdout, self.previous_stderr
        self.event('session_closed')

if 'report' in globals():
    report.close()
ROLE = 'M5'
report = ReportSession(PROJECT_ROOT, ROLE, DATA_PATH, df)
ARTIFACT_PATHS = report.paths
OUTPUTS = {}  # Add only results actually produced during this run.
report.upstream()
inputs = {key: report.load(key) for key in UPSTREAM_KEYS[ROLE]}
print('Path รวมทั้งหมด: ARTIFACT_PATHS หรือ artifact_paths.json; ไม่ต้องรอทุกไฟล์ก่อนเริ่ม')
try:
    shell = get_ipython()
except NameError:
    shell = None
if shell is not None:
    report.install_capture(shell)
print('Loaded:', DATA_PATH.name, '| rows:', len(df), '| columns:', len(df.columns))
```

### เทียบ Colab กับแอป

```python
import importlib.util
import sys
import subprocess
if importlib.util.find_spec('streamlit') is None:
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', 'streamlit'])
from streamlit.testing.v1 import AppTest
namespace = {'__name__': 'm5_validation'}
exec(compile(app_source, str(app_path), 'exec'), namespace)
validate = namespace['validate_inputs']
valid = dict(hour=12, temperature_c=25.0, humidity_pct=70.0, weather='Clear', wind_speed_kmh=10.0, is_working_day=1, day_of_week=2)
records = []
def check(case, action, expected_error=False):
    try:
        action()
        passed = not expected_error
        actual = 'accepted'
    except ValueError as error:
        passed = expected_error
        actual = str(error)
    records.append(dict(case=case, expected='reject' if expected_error else 'accept', actual=actual, passed=bool(passed)))
check('normal input', lambda: validate(valid))
check('hour boundary 0', lambda: validate({**valid, 'hour': 0}))
check('hour boundary 23', lambda: validate({**valid, 'hour': 23}))
check('invalid hour 24', lambda: validate({**valid, 'hour': 24}), True)
check('missing weather', lambda: validate({k:v for k,v in valid.items() if k != 'weather'}), True)
check('unknown weather', lambda: validate({**valid, 'weather':'Snow'}), True)
check('negative wind', lambda: validate({**valid, 'wind_speed_kmh':-1}), True)
check('nonfinite temperature', lambda: validate({**valid, 'temperature_c':float('nan')}), True)
check('string temperature', lambda: validate({**valid, 'temperature_c':'hot'}), True)
check('weekday consistency', lambda: validate({**valid, 'day_of_week':6}), True)

final_pipeline = load('final_pipeline')
for weather in ['Clear','Cloudy','Light_Rain','Heavy_Rain']:
    at = AppTest.from_file(str(app_path)).run(timeout=30)
    assert len(at.exception)==0
    at.selectbox[0].select('Wednesday')
    at.selectbox[1].select(weather)
    at.button[0].click().run(timeout=30)
    expected_prediction = float(final_pipeline.predict(validate({**valid,'weather':weather}))[0])
    actual_prediction = float(at.session_state['last_prediction'])
    records.append(dict(case='parity '+weather,expected=str(expected_prediction),actual=str(actual_prediction),passed=abs(expected_prediction-actual_prediction)<=1e-9 and len(at.exception)==0 and len(at.error)==0))
at.selectbox[0].select('Sunday').run()
records.append(dict(case='Sunday UI',expected='working day No',actual=str([c.value for c in at.caption]),passed=any('Working day: No' in c.value for c in at.caption)))
at.number_input[0].set_value(23)
at.button[0].click().run(timeout=30)
expected_prediction=float(final_pipeline.predict(validate({**valid,'hour':23,'day_of_week':6,'is_working_day':0,'weather':'Heavy_Rain'}))[0])
actual_prediction=float(at.session_state['last_prediction'])
records.append(dict(case='Sunday hour23 parity',expected=str(expected_prediction),actual=str(actual_prediction),passed=abs(expected_prediction-actual_prediction)<=1e-9 and len(at.exception)==0))
app_test_log = pd.DataFrame(records)
save('app_test_log',app_test_log)
print(app_test_log.to_string(index=False))
assert app_test_log.passed.all(), 'Resolve failed application tests'
assert sha(PATHS['final_pipeline'])==load('final_test_metrics')['pipeline_hash']==load('model_metadata')['pipeline_sha256']
print('APP TESTS PASS',len(app_test_log),'/',len(app_test_log))
print('Artifact schema and all three model hashes match.')
status='Corrected artifacts and app passed '+str(len(app_test_log))+' validation/parity tests on hosted Colab. Final test was repeated after repair and is not a fresh untouched holdout. Deployment package prepared; not yet deployed. Monitoring requires actual live observations.'
(OUTPUTS_DIR/'M5_HANDOFF_STATUS.md').write_text(status,encoding='utf-8')
(REPAIR_DIR/'app_test_log.csv').write_text(app_test_log.to_csv(index=False),encoding='utf-8')
```
